```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.model_selection import train_test_split

# Dataset proprio - Agencia de Viagens
dados_viagem = {
    'mensagem': [
        'Quero comprar uma passagem para o Rio de Janeiro',
        'Gostaria de reservar um voo para Salvador',
        'Como faço para comprar uma passagem aerea?',
        'Quero passagem de ida e volta para Recife',
        'Preciso comprar um bilhete para São Paulo',
        'Quero cancelar minha reserva de viagem',
        'Como cancelo a passagem que comprei ontem?',
        'Preciso desmarcar meu voo para amanha',
        'Gostaria de cancelar a reserva do hotel',
        'Quero falar com um atendente humano',
        'Preciso de ajuda de uma pessoa de verdade',
        'Pode me transferir para o suporte, por favor?'
    ],
    'intencao': [
        'comprar_passagem', 'comprar_passagem', 'comprar_passagem', 'comprar_passagem', 'comprar_passagem',
        'cancelar_reserva', 'cancelar_reserva', 'cancelar_reserva', 'cancelar_reserva',
        'falar_atendente', 'falar_atendente', 'falar_atendente'
    ]
}

df4 = pd.DataFrame(dados_viagem)

X = df4['mensagem']
y = df4['intencao']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

vectorizer = TfidfVectorizer()
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)

modelo_nb = MultinomialNB()
modelo_nb.fit(X_train_vec, y_train)

frases_novas = [
    "Quero marcar uma passagem para Fortaleza",
    "Preciso cancelar minha viagem de fim de semana",
    "Queria falar com uma pessoa do atendimento"
]

frases_novas_vec = vectorizer.transform(frases_novas)
predicoes = modelo_nb.predict(frases_novas_vec)

print("--- RESULTADOS DO LAB 04 ---")
for frase, intencao in zip(frases_novas, predicoes):
    print(f"Mensagem: '{frase}' ==> Intenção Predita: [{intencao}]")
```
